# Project: Data-Driven Growth Strategy for a Retail Brand

This code generates a synthetic retail dataset designed for strategic business analysis. The resulting database consists of **4 interconnected tables**, simulating a real-world retail ecosystem:

* **Customers Table:** Contains unique identifiers, demographic details, and registration dates.
* **Products Table:** Includes a hierarchical structure of categories and product names with associated price points.
* **Orders Table:** A transactional table linking customers to products, featuring specific order dates, quantities, and order IDs.
* **Campaigns Table:** Contains marketing data used to track the effectiveness of different growth initiatives.

## 1. Environment Setup & Initialization

This cell initializes the project environment by importing the necessary libraries and setting a fixed seed for random number generation.

* **NumPy (`np`)**: Used for high-performance mathematical operations and random sampling.
* **Pandas (`pd`)**: The primary tool for data manipulation and creating the final structured DataFrames.
* **google.colab (`files`)**: Utility for downloading the generated datasets locally from the Google Colab environment.
* **Random Seed (`42`)**: Ensures that every time this code is run, it generates the **exact same** "random" data. This is crucial for consistent analysis and debugging.

```python
import numpy as np
import pandas as pd
from google.colab import files
from numpy._core.fromnumeric import choose
np.random.seed(42)
```

## 2. Generating the Customer Dataset

In this cell, we create the core **Customer Table**, representing our user base of 100 unique individuals. The script simulates realistic demographic and behavioral traits.

### Key Logic & Attributes:
* **Customer_ID**: Uses `zfill(4)` to create standardized, sortable identifiers (e.g., `CUST_0001`).
* **Demographics**: Randomly distributes **Age** (18–80) and **City** across five major US metropolitan hubs.
* **Signup_Date**: Utilizes `normalize()` to generate clean timestamps between 2020 and 2025, stripping away hours and minutes for daily-level analysis.
* **Acquisition_Channel**: Tracks how the customer found the brand (e.g., Social Media vs. Direct), which is vital for calculating **Marketing Attribution**.



### Technical Validation:
The cell concludes by printing the DataFrame **shape** and **data types** to verify that the dates were correctly converted and that the dataset contains exactly 100 entries.

```python
n = 100
city = ['New York', 'Los Angeles','Chicago','Houston','Philadelphia']
acquisition_channels = ['search engine ','paid ads','social media','direct']

customers = pd.DataFrame({
 "Customer_ID": [f"CUST_{str(i).zfill(4)}" for i in range(1,n+1)],
 "Age": np.random.randint(18,80,n),
 "Signup_Date": pd.to_datetime(
     np.random.randint(
         pd.Timestamp("2020-01-01").value,
         pd.Timestamp("2025-12-31").value,
         n)).normalize(),
 "City": np.random.choice(city,n),
 "Acquisition_Channel": np.random.choice(acquisition_channels,n)
 })


print(f"type of customers : {type(customers)}")
print(f"\nshape: {customers.shape}")
print(f"\nData types :\n{customers.dtypes}")
customers.head(10)
```

## 3. Creating the Product Catalog

This cell builds the **Products Table**, which serves as the master catalog for the retail brand. The key challenge here is maintaining **logical consistency**: ensuring that a "Smart Watch" is always categorized under "Electronics" and priced within a realistic range.

### Key Logic & Features:
* **Hierarchical Mapping**: Uses a dictionary to link specific product names to their respective categories. This ensures that the dataset doesn't accidentally assign "Olive Oil" to the "Electronics" category.
* **Category-Based Pricing**: Instead of random prices, the script uses a `price_ranges` dictionary to apply logic-based pricing. For example, a **Book** will typically cost between $8 and $50, while **Electronics** can go up to $500.
* **Uniform Distribution**: `np.random.uniform` is used to create realistic price variation, and `round(..., 2)` ensures the values look like actual currency.
* **Scalable IDs**: Uses `PROD_0001` format, allowing for a standardized catalog that remains easy to sort and index.



### Business Value:
By establishing clear categories and price points, this table enables **Margin Analysis** and **Category Performance** tracking. It allows the growth strategy to identify which product segments are driving the most revenue.

```python
n= 50
categories = ["Electronics", "Clothing", "Home & Kitchen", "Sports", "Books", "Beauty", "Toys", "Food & Beverage"]
product_names = {
    "Electronics":      ["Wireless Headphones", "Bluetooth Speaker", "Smart Watch", "Laptop Stand", "USB Hub"],
    "Clothing":         ["Running Jacket", "Denim Jeans", "Cotton T-Shirt", "Wool Sweater", "Sports Shorts"],
    "Home & Kitchen":   ["Air Fryer", "Coffee Maker", "Blender", "Knife Set", "Cutting Board"],
    "Sports":           ["Yoga Mat", "Resistance Bands", "Dumbbells", "Jump Rope", "Water Bottle"],
    "Books":            ["Python for Beginners", "Data Science Handbook", "The Art of War", "Atomic Habits", "Clean Code"],
    "Beauty":           ["Face Moisturizer", "Lip Balm", "Shampoo", "Sunscreen SPF50", "Eye Cream"],
    "Toys":             ["Building Blocks", "Remote Car", "Puzzle Set", "Action Figure", "Board Game"],
    "Food & Beverage":  ["Green Tea", "Protein Powder", "Olive Oil", "Honey Jar", "Mixed Nuts"],
}

#sample categories
sampled_categories = np.random.choice(categories,n)
#match product names to categories
sampled_names = [np.random.choice(product_names[cat]) for cat in sampled_categories ]

# price ranges per category
price_ranges = {
    "Electronics":      (20, 500),
    "Clothing":         (10, 150),
    "Home & Kitchen":   (15, 300),
    "Sports":           (5,  200),
    "Books":            (8,   50),
    "Beauty":           (5,  100),
    "Toys":             (10, 120),
    "Food & Beverage":  (3,   60),
}

# sampled prices  to mathc categories
sampled_prices = [round(np.random.uniform(*price_ranges[cat]),2)for cat in sampled_categories]

products = pd.DataFrame({
    "Product_ID": [f"PROD_{str(i).zfill(4)} "for i in range(1,n+1)],
    "Product_Name":sampled_names,
    "Category": sampled_categories,
    "Price": sampled_prices
})

print(f"type of orders :{type(products)}")
print(f"\nshape : {products.shape}")
print(f"\nData types{products.dtypes}")
products.head(10)
```

## 4. Generating the Transactional Orders Table

This cell is the "engine" of the dataset, where we simulate the actual purchasing behavior by connecting customers, products, and marketing campaigns. It transforms our high-level logic into 5,000 specific transaction rows.

### Key Logic & Constraints:
* **Controlled Distribution**: Uses a "Fill and Top-up" algorithm to ensure exactly 5,000 rows are distributed across 780 orders, with each order containing a realistic count of **3 to 7 items**.
* **Unique Shopping Baskets**: Within a single order, the script uses `replace=False` when picking products. This ensures a customer doesn't have duplicate rows for the same item in one transaction.
* **Transactional Integrity**: Every order is assigned a `Campaign_ID`, allowing us to trace which marketing effort drove the sale.

> ### ⚠️ Critical Business Logic: Temporal Consistency
> A major feature of this cell is the **Date Validation Logic**. To maintain a realistic timeline, the script ensures that **an Order Date can never occur before a Customer's Signup Date**.
> 1. It uses a **Lookup Dictionary** (`reg_date_lookup`) to find the specific registration date for each customer.
> 2. It calculates the available `days_range` between that signup date and the simulation end date.
> 3. It generates an `order_date` within that specific window, preventing "time-travel" errors in the data.

```python
n_orders = 780
total_rows   = 5000
min_items , max_items = 3,7
customer_ids = [f"CUST_{str(i).zfill(4)}"for i in range(1,101)]
product_ids  = [f"PROD_{str(i).zfill(4)}"for i in range(1,51) ]
campaign_ids = [f"CAMP_{str(i).zfill(3)}"for i in range(1,21) ]

# set 3 items per order first , which means we have around 2430 rows
itmes_per_order = np.full(n_orders, min_items,dtype= int)
# the remaining rows 2660
remaining  = total_rows - (min_items * n_orders )

# let's randomly add extra row but not exceeding the max_items , so it will be 4 maximum
while remaining > 0 :
  eligible = np.where(itmes_per_order < max_items)[0]
  pick = np.random.choice(eligible)
  itmes_per_order[pick] += 1
  remaining -= 1

# before creting orders table that will contain customer id columns and purchase date of orders , this date can't be before signup date of customer
# lookup dictionary to get sign up date for each customer
reg_date_lookup = customers.set_index("Customer_ID")["Signup_Date"].to_dict()

# build orders
rows =[]
max_order_date = pd.Timestamp("2025-12-31")
for order_num in range(n_orders):
  order_id = f"ORD_{str(order_num+1).zfill(5)}"
  choosen_customer_id = np.random.choice(customer_ids)
  campaing_id = np.random.choice(campaign_ids)
  n_items = itmes_per_order[order_num]

  reg_date = pd.Timestamp(reg_date_lookup[choosen_customer_id])
  days_range = max(1, (max_order_date - reg_date).days)
  order_date = reg_date + pd.Timedelta(days = int(np.random.randint(1,days_range+1)))


  selected_products = np.random.choice(product_ids,n_items,replace = False)
  for product_id in selected_products :
    rows.append({
      "Order_ID": order_id,
      "Customer_ID" : choosen_customer_id,
      "Order_Date": order_date,
      "Product_ID": product_id,
      "Quantity" : int(np.random.randint(1,5)),
      "Campaign_ID": campaing_id
    })
#create dataframe
orders = pd.DataFrame(rows , columns =["Order_ID","Customer_ID","Order_Date","Product_ID","Quantity","Campaign_ID"])

print(orders.head(10))
print(f"Totla rows{len(orders)}")
print(f"\nUnique Orders {orders['Order_ID'].nunique()}")
print(f"\nUnique customers{orders['Customer_ID'].unique()}")
print(f"\nUniqy Products{orders['Product_ID'].unique()}")
print(f"\nData types\n {orders.dtypes}")
print(orders[orders["Order_ID"]=="ORD_00001"])
```

## 5. Building the Campaigns Table (The "Better Late Than Never" Edition)

This final cell creates the **Campaigns Table**, defining the marketing initiatives that allegedly drove our sales. Here, we categorize our marketing efforts into "Control" groups and different "Discount" variations to enable A/B testing analysis.

### Logic & "Creative" Timing:
* **A/B Testing Structure**: Assigns each campaign to a `Test_Group` (Control, Discount_A, or Discount_B), which is the bread and butter of growth strategy experiments.
* **Retroactive Reality**: Since we decided to live life on the edge and build the **Orders Table first**, we can't just pick random campaign dates and hope for the best.
* **The "Reverse Fix"**: To avoid the embarrassment of having customers buy things through a campaign that hasn't started yet, the script looks at the actual sales data (`min` and `max` order dates) and stretches the campaign dates to fit around them.

> ### 🙄 A Note on Workflow
> In a perfect, organized world, you would build your Campaign dates **before** your Orders. That way, the marketing starts, and *then* the sales happen. But since we already coded the Orders table and were too far gone to turn back, we’ve used this "Time-Travel Logic" to force the Campaigns to be compatible with the Orders we already created. It’s not "backward thinking"—it’s "agile pivoting," haha!

```python

used_campaign_ids = orders["Campaign_ID"].unique()
# for each campagin get min and max order date
campaign_order_date =(
    orders.groupby("Campaign_ID")["Order_Date"]
    .agg(first_order = "min",last_order = "max")
    .reset_index()
)

# build camapaign table
test_group= ["Control", "Discount_A", "Discount_B"]
rows = []
for _,row in campaign_order_date.iterrows():
  first_order = pd.Timestamp(row["first_order"])
  last_order  = pd.Timestamp(row["last_order"])

  # camapaign starts randomly before first order by 1-30 days
  start_offset = int(np.random.randint(1, 31))
  start_date   = first_order - pd.Timedelta(days=start_offset)

  # camapaign starts randomly before first order by 1-30 days
  end_offset = int(np.random.randint(1, 31))
  end_date   = last_order + pd.Timedelta(days=end_offset)

  rows.append({
        "Campaign_ID": row["Campaign_ID"],
        "Test_Group":  np.random.choice(test_group),
        "Start_Date":  start_date,
        "End_Date":    end_date,
    })

campaigns = pd.DataFrame(rows, columns=["Campaign_ID", "Test_Group", "Start_Date", "End_Date"])



print(f"Shape: {campaigns.shape}")
print(f"\nData Types:\n{campaigns.dtypes}")
campaigns
```

# export and download data

```python
dataframes = {
    "customers": customers,
    "products": products,
    "orders": orders,
    "campaigns": campaigns
}

for name, df in dataframes.items():
  df.to_csv(f"{name}.csv",index=False)
  print(f"{name}.csv saved - {df.shape[0]} rows X {df.shape[1]} cols")


for name in dataframes.keys():
  files.download(f"{name}.csv")
  print(f"Downloading {name}.csv")
```
